# House Sales in King County, USA


This dataset contains house sale prices for King County, which includes Seattle. It includes homes sold between May 2014 and May 2015.


| Variable      | Description                                                                                                 |
| ------------- | ----------------------------------------------------------------------------------------------------------- |
| id            | A notation for a house                                                                                      |

| price         | Price is prediction target                                                                                  |
| bedrooms      | Number of bedrooms                                                                                          |
| bathrooms     | Number of bathrooms                                                                                         |
| sqft_living   | Square footage of the home                                                                                  |
| sqft_lot      | Square footage of the lot                                                                                   |
| floors        | Total floors (levels) in house                                                                              |
| waterfront    | House which has a view to a waterfront                                                                      |
| view          | Has been viewed                                                                                             |
| condition     | How good the condition is overall                                                                           |
| grade         | overall grade given to the housing unit, based on King County grading system                                |

| sqft_basement | Square footage of the basement                                                                              |
| yr_built      | Built Year                                                                                                  |

| lat           | Latitude coordinate                                                                                         |

| sqft_living15 | Living room area in 2015(implies-- some renovations) This might or might not have affected the lotsize area |

### Question 4

Use the function <code>boxplot</code> in the seaborn library  to  determine whether houses with a waterfront view or without a waterfront view have more price outliers.


```python
sns.boxplot(x='waterfront', y='price', data=df)
plt.title('Price Distribution by Waterfront View')
plt.xlabel('Waterfront View (0: No, 1: Yes)')
plt.ylabel('Price')
plt.show()
```

The boxplot shows the distribution of house prices for houses with and without a waterfront view.

The box on the left (labeled 0) represents houses without a waterfront view. It shows that most houses without a waterfront view have prices concentrated in the lower range, with many outliers extending to much higher prices.
The box on the right (labeled 1) represents houses with a waterfront view. This box is higher on the price scale, indicating that houses with a waterfront view generally have higher prices. It also shows outliers, but the overall price range and the median price are significantly higher compared to houses without a waterfront view.
Based on the plot, houses without a waterfront view appear to have more price outliers in terms of the sheer number of individual points plotted above the upper whisker. However, houses with a waterfront view have a higher median price and a larger interquartile range, indicating a wider spread of prices within the main body of the data for that group.

### Question 5

Use the function <code>regplot</code>  in the seaborn library  to  determine if the feature <code>sqft_above</code> is negatively or positively correlated with price.


```python
sns.regplot(x='sqft_above', y='price', data=df, line_kws={"color": "red"})
plt.title('Relationship between sqft_above and Price')
plt.xlabel('Square Footage Above Ground')
plt.ylabel('Price')
plt.show()
```

The upward slope of the red line indicates a positive correlation between sqft_above and price. This means that as the square footage above ground increases, the price of the house tends to increase as well. The scattering of the points around the line suggests that while there is a positive correlation, other factors also influence the price.

We can use the Pandas method <code>corr()</code>  to find the feature other than price that is most correlated with price.


```python
df.drop('date', axis=1).corr()['price'].sort_values()
```

Based on the correlation values I calculated, the features most positively correlated with price are:

* sqft_living: 0.702035
* grade: 0.667434
* sqft_above: 0.605567
* sqft_living15: 0.585379
* bathrooms: 0.525738

These features have the highest positive correlation coefficients with the price, indicating that as their values increase, the price tends to increase as well.

# Module 5: Model Evaluation and Refinement


Import the necessary modules:


```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
print("done")
```

We will split the data into training and testing sets:


```python
features =["floors", "waterfront","lat" ,"bedrooms" ,"sqft_basement" ,"view" ,"bathrooms","sqft_living15","sqft_above","grade","sqft_living"]
X = df[features]
Y = df['price']

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.15, random_state=1)


print("number of test samples:", x_test.shape[0])
print("number of training samples:",x_train.shape[0])
```

### Question 9

Create and fit a Ridge regression object using the training data, set the regularization parameter to 0.1, and calculate the R^2 using the test data.


```python
from sklearn.linear_model import Ridge
```

```python
ridge_model = Ridge(alpha=0.1)
ridge_model.fit(x_train, y_train)
r2_test = ridge_model.score(x_test, y_test)
print('R^2 on test data:', r2_test)
```

### Question 10

Perform a second order polynomial transform on both the training data and testing data. Create and fit a Ridge regression object using the training data, set the regularisation parameter to 0.1, and calculate the R^2 utilising the test data provided. Take a screenshot of your code and the R^2.


```python
from sklearn.preprocessing import PolynomialFeatures

# Create a PolynomialFeatures object with degree 2
poly = PolynomialFeatures(degree=2)

# Transform the training and testing data
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

# Create and fit a Ridge regression object
ridge_model = Ridge(alpha=0.1)
ridge_model.fit(x_train_poly, y_train)

# Calculate the R^2 using the test data
r2_test_poly = ridge_model.score(x_test_poly, y_test)
print('R^2 on test data with polynomial features:', r2_test_poly)
```
